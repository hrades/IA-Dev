3. Dividir os dados em conjunto de treinamento e teste e fazer a regressão com pelo menos três modelos diferentes

Prompt utilizado no código abaixo:  
Faça a padronização da base de dados, que está na variável 'dados', das colunas distance_to_school, distance_to_supermarket_km e crime_rate_index.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Separar features e target
X = dados.drop('property_value', axis=1)
y = dados['property_value']

# Colunas a serem padronizadas
columns_to_standardize = ['distance_to_school', 'distance_to_supermarket_km', 'crime_rate_index']

# Criar uma cópia do DataFrame para evitar SettingWithCopyWarning
X_scaled = X.copy()

# Padronizar as colunas específicas
scaler = StandardScaler()
X_scaled[columns_to_standardize] = scaler.fit_transform(X_scaled[columns_to_standardize])

# Dividir os dados em conjuntos de treino e teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("Dados divididos em conjuntos de treino e teste:")
print("Formato de X_train:", X_train.shape)
print("Formato de X_test:", X_test.shape)
print("Formato de y_train:", y_train.shape)
print("Formato de y_test:", y_test.shape)
```

Prompt para o código abaixo:  
Treine um modelo de regressão linear com o conjunto de dados de treino e teste que foram criados (X_train, X_test, y_train, y_test). Adicione a previsão e avalie a confiabilidade do modelo, utilizando mse, rmse e r²

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Treinar o modelo de Regressão Linear
model_lr = LinearRegression()
model_lr.fit(X_train, y_train)

# Fazer previsões no conjunto de teste
y_pred_lr = model_lr.predict(X_test)

# Avaliar o modelo
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("Métricas de desempenho para Regressão Linear:")
print("MSE:", mse_lr)
print("RMSE:", rmse_lr)
print("R²:", r2_lr)
```

Prompt para o código abaixo:  
Treine um modelo de árvore aleatória com o conjunto de dados de treino e teste que foram criados (X_train, X_test, y_train, y_test). Adicione a previsão e avalie a confiabilidade do modelo, utilizando mse, rmse e r?

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Treinar o modelo de Random Forest
model_rf = RandomForestRegressor(random_state=42)
model_rf.fit(X_train, y_train)

# Fazer previsões no conjunto de teste
y_pred_rf = model_rf.predict(X_test)

# Avaliar o modelo
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("Métricas de desempenho para Random Forest:")
print("MSE:", mse_rf)
print("RMSE:", rmse_rf)
print("R²:", r2_rf)
```

Prompt do código abaixo:  
Treine um modelo de Extreme Gradient Boosting com o conjunto de dados de treino e teste que foram criados (X_train, X_test, y_train, y_test). Adicione a previsão e avalie a confiabilidade do modelo, utilizando mse, rmse e r²

```python
import xgboost as xgb
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Treinar o modelo de Extreme Gradient Boosting (XGBoost)
model_xgb = xgb.XGBRegressor(objective='reg:squarederror', random_state=42)
model_xgb.fit(X_train, y_train)

# Fazer previsões no conjunto de teste
y_pred_xgb = model_xgb.predict(X_test)

# Avaliar o modelo
mse_xgb = mean_squared_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mse_xgb)
r2_xgb = r2_score(y_test, y_pred_xgb)

print("Métricas de desempenho para Extreme Gradient Boosting (XGBoost):")
print("MSE:", mse_xgb)
print("RMSE:", rmse_xgb)
print("R²:", r2_xgb)
```

4. Metricas de desempenho

(Usar as métricas de desempenho para comparar os resultados)

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error

# Função para avaliar o modelo usando validação cruzada
def evaluate_model_cv(model, X, y):
    mse_scores = cross_val_score(model, X, y, scoring='neg_mean_squared_error', cv=10)
    rmse_scores = np.sqrt(-mse_scores)
    r2_scores = cross_val_score(model, X, y, scoring='r2', cv=10)
    return -mse_scores.mean(), rmse_scores.mean(), r2_scores.mean() # Return MSE, RMSE, R²

# Avaliar Regressão Linear com validação cruzada
mse_lr_cv, rmse_lr_cv, r2_lr_cv = evaluate_model_cv(model_lr, X, y)
print(f'Regressão Linear (Validação Cruzada)')
print(f'Mean Squared Error (MSE): {mse_lr_cv}')
print(f'Root Mean Squared Error (RMSE): {rmse_lr_cv}')
print(f'R-squared (R²): {r2_lr_cv}')
print("-" * 30)

# Avaliar Árvore Aleatória com validação cruzada
mse_rf_cv, rmse_rf_cv, r2_rf_cv = evaluate_model_cv(model_rf, X, y)
print(f'Árvore Aleatória (Validação Cruzada)')
print(f'Mean Squared Error (MSE) [Random Forest]: {mse_rf_cv}')
print(f'Root Mean Squared Error (RMSE) [Random Forest]: {rmse_rf_cv}')
print(f'R-squared (R²) [Random Forest]: {r2_rf_cv}')
print("-" * 30)

# Avaliar Extreme Gradient Boosting com validação cruzada
mse_xgb_cv, rmse_xgb_cv, r2_xgb_cv = evaluate_model_cv(model_xgb, X, y)
print(f'Extreme Gradient Boosting (Validação Cruzada)')
print(f'Mean Squared Error (MSE): {mse_xgb_cv}')
print(f'Root Mean Squared Error (RMSE): {rmse_xgb_cv}')
print(f'R-squared (R²): {r2_xgb_cv}')
```
